```python
import pandas as pd
df = pd.DataFrame({
'country': ['India','USA','Germany','Niger','Afghanistan'],
'child_mort': [45,6,4,170,78],
'income': [6000,50000,44000,800,1500],
'life_expec': [69,78,81,58,60],
'health': [4,17,11,3,8],
'gdpp': [2000,60000,45000,400,1000],
'exports': [40,15,50,10,10],
'imports': [50,20,40,60,50],
'inflation': [6,2,1,3,5]
})
df['Trade_Balance'] = df['exports'] - df['imports']
print(df)

# Inflation fix
df['inflation'] = df['inflation'].clip(lower=0)

df['Development_Index'] = (df['income'] + df['gdpp'] + df['life_expec']) / df['child_mort']
df['Health_Impact_Ratio'] = df['health'] / df['child_mort']

def seg(r):
    if r['child_mort'] > 80:
        return 'High Risk Country'
    elif r['income'] > 30000:
        return 'Developed Nation'
    else:
        return 'Moderate Risk'

df['segment'] = df.apply(seg, axis=1)

print("Step 2 SUCCESS!")
print(df[['country','segment','Trade_Balance','Development_Index']])

df.to_csv('clean_country_data.csv', index=False)
print("File saved - clean_country_data.csv")
```
