```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import streamlit as st

st.title("Country Economic Metrics Analysis")

df = pd.read_csv("processed_data.csv")

st.write("### Data Summary", df.head())

# Dropdown selection for metric to plot
metric = st.selectbox(
    "Select Metric to Plot against Income",
    ["child_mort", "inflation", "life_expec", "health"]
)

fig, ax = plt.subplots()
sns.regplot(
    data=df,
    x="income",
    y=metric,
    ax=ax,
    scatter_kws={'s':100}
)
ax.set_title(f"Income vs {metric.replace('_', ' ').title()}")

st.pyplot(fig)
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import streamlit as st

st.title("Global Development Data Dashboard")

df = pd.read_csv("processed_data.csv")

st.write("### Sample Data Preview", df.head())

# Dropdown to filter by development segment
selected_segment = st.selectbox(
    "Select Development Segment",
    df["segment"].unique()
)

filtered_df = df[
    df["segment"] == selected_segment
]

# Visualize Life Expectancy vs GDP per Capita
fig, ax = plt.subplots()
sns.scatterplot(
    data=filtered_df,
    x="gdpp",
    y="life_expec",
    hue="country",
    size="income",
    sizes=(50, 200),
    ax=ax
)
ax.set_title("GDP per Capita vs Life Expectancy")
ax.set_xlabel("GDP per Capita (gdpp)")
ax.set_ylabel("Life Expectancy")

st.pyplot(fig)
```

```python
import sqlite3

# Connect to an in-memory SQLite database
# To create a persistent file-based database, change ':memory:' to a filename like 'my_database.db'
conn = sqlite3.connect(':memory:')
cursor = conn.cursor()

create_table_sql = """
CREATE TABLE country_development (
    country VARCHAR(100) PRIMARY KEY,
    child_mort FLOAT,
    exports FLOAT,
    health FLOAT,
    imports FLOAT,
    income FLOAT,
    inflation FLOAT,
    life_expec FLOAT,
    total_fer FLOAT,
    gdpp FLOAT,
    segment VARCHAR(50)
);
"""

try:
    cursor.execute(create_table_sql)
    conn.commit()
    print("Table 'country_development' created successfully.")
except sqlite3.Error as e:
    print(f"An error occurred: {e}")
finally:
    conn.close()
```
